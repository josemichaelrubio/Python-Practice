# defaultdict

Found in Python's built-in collections module

works like a standard dict with one major difference: it never raises a `KeyError`
- KeyError. If you try to access or modify a key that doesn't exist in a regular dictionary, Python throws an error.

A defaultdict solves this by automatically creating the missing key and assigning it a default value on the fly.

When you create a defaultdict, you pass it a callable argument (like a function or a data type) known as a `default_factory`. Whenever you ask for a key that isn't already in the dictionary, the defaultdict calls this factory to generate the initial value.

```python
from collections import defaultdict

student_grades = defaultdict(list)

grades = [('Alice', 'A'), ('Bob', 'B'), ('Alice', 'A-'), ('Charlie', 'C')]

for name, grade in grades:
    student_grades[name].append(grade)

print(dict(student_grades))
#print(student_grades)
```

```python
from collections import defaultdict

# Use a lambda to return a specific string for missing keys
user_roles = defaultdict(lambda: "Guest")

user_roles['Alice'] = "Admin"

print(user_roles['Alice'])  # Output: Admin
print(user_roles['Bob'])    # Output: Guest (Bob didn't exist!)
```

```python
from collections import defaultdict

# Grouping unique classes a student is taking
student_classes = defaultdict(set)
student_classes['Alice'].add('Math')
student_classes['Alice'].add('Math') # Sets ignore duplicates
student_classes['Alice'].add('Science')

print(student_classes['Alice']) 
# Output: {'Math', 'Science'}
```
