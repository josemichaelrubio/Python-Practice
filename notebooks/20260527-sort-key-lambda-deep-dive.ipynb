Given an array of intervals where intervals[i] = [start, end], merge all overlapping intervals, and return an array of the non-overlapping intervals that cover all the intervals in the input.

```python
## Expected input
intervals = [[1, 3], [2, 6], [8, 10], [15, 18]]

## Expected output
# [[1,6],[8,10],[15,18]]
```

*How are can we make this linear. Cause if intervals are sorted by their star times, the problems simplifies signifcantly*
- Once sorted you never need to compare the current interval with the last merged interval you've processed
 

```python
# Sort the input array by the starting element of each interval
def merge_intervals(intervals):
    # Guard Clause, different than a Base base cause its for loops
    if not intervals:
        return []
    
    #1)intervals.sort() sorts the list itself
    #2)key= expects a function, and python will pass every single item in the list thorugh this function before making comparision. Whatever it returns, python use to tdermine the sorting order
        # key allows to sort in a specific way, cause by default, sort() sorts by smallest to largestla
    #3)lambda x: x[0] is an anonymous function (a one-line function without a name)
        # x represents a single item  from the list, like one inteval [2,6]
        # x[0] return the value (the start time, 2)
        # thus lambda arguements : expressions
        # this lambda sorts by the first element of given list
            # This is done by default, but if there is a tie breaker, it would then look at the 2nd element of the list
                # this can cause a TypeError if the 2nd element if not element that cant be sorted like an element like an object
    intervals.sort(key=lambda x: x[0])

    merged = []
    for interval in intervals:
        if not merged or merged[-1][1] < interval[0]:
            merged.append(interval)
        else:
            merged[-1][1] = max(merged[-1][1], interval[1])
    return merged

output = merge_intervals(intervals)
print(output)
```

- intervals.sorted() sorts the list in place, modifies the original list directly and returns none
- sorted(intervals) returns a brand new list
