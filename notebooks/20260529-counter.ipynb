Counter from collections module is a built-in tool designed specificly for counting hashable objects

the item is stored as a diectionary keys and their counts are stored as dictionary values.

It replaces loops that check if an item exist then increment its count. Very optimized

.most_common()


```python
from collections import Counter

c = Counter("banana")   # {'a':3,'n':2,'b':1}
c = Counter([1,1,2,3])  # {1:2, 2:1, 3:1}

c.most_common(2)        # [(1, 2), (2, 1)]
c['z']                  # 0 — never KeyError

# combine — using real Counters
a = Counter("aab")      # {'a':2, 'b':1}
b = Counter("ab")       # {'a':1, 'b':1}

a - b   # Counter({'a': 1})         — subtracts, drops zeros
a + b   # Counter({'a': 3, 'b': 2}) — merges
a & b   # Counter({'a': 1, 'b': 1}) — intersection (min of each)
a | b   # Counter({'a': 2, 'b': 1}) # union (max of each)
```

```python
from typing import List
from collections import Counter

nums = [1, 2, 3, 3]

from collections import Counter
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        c = Counter(nums)
        for i in c.values():
            if i > 1:
                return True
        return False

solution = Solution()

print(solution.hasDuplicate(nums))
```

```python
nums = [1, 2, 3, 3]

class Solution:
    def hasDuplicate(self, nums: list[int]) -> bool:
        seen = set()
        for num in nums:
            if num in seen:
                return True
            seen.add(num)
        return False

solution = Solution()

print(solution.hasDuplicate(nums))
```

```python
from collections import Counter

s = "racecar"
t = "carrace"

class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False
        return Counter(s) == Counter(t)
    

solution = Solution()

print(solution.isAnagram(s,t))
```

In Python, when you use the == operator to compare two dictionaries (or Counter objects), Python does not care about the order of the items. It only checks two things:
1. Do both dictionaries have the exact same keys (letters)?
2. Do those keys have the exact same values (counts)?
