```python
import collections

def isValidSudoku(board):
    # Create dictionaries where each key automatically maps to an empty Hash Set
    cols = collections.defaultdict(set)
    rows = collections.defaultdict(set)
    squares = collections.defaultdict(set)  # Key will be a tuple: (r // 3, c // 3)

    for r in range(9): # start with the first row
        for c in range(9): # then complete the first row's columns before going to the 2nd row
            # value
            val = board[r][c]
            
            # Ignore empty spaces
            if val == ".":
                continue
            
            # Rule check: Is the value already in our sets?
            if (val in rows[r] or 
                val in cols[c] or 
                val in squares[(r // 3, c // 3)]):
                return False
            
            # Record the value in our sets for future checks
            rows[r].add(val)
            cols[c].add(val)
            squares[(r // 3, c // 3)].add(val)

    # If we finish the loop without conflicts, the board is valid
    return True
```
