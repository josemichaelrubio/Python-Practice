```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        # Set our two pointers at the start and end of the string
        left = 0
        right = len(s) - 1

        # Keep checking as long as the left pointer is before the right pointer
        while left < right:
            
            # Move the left pointer forward if it's pointing to a space or punctuation
            while left < right and not s[left].isalnum():
                left += 1
                
            # Move the right pointer backward if it's pointing to a space or punctuation
            while left < right and not s[right].isalnum():
                right -= 1
                
            # Compare the characters (converted to lowercase so 'A' == 'a')
            if s[left].lower() != s[right].lower():
                return False # Mismatch found, not a palindrome!
            
            # If they match, move both pointers inward for the next check
            left += 1
            right -= 1
            
        # If the loop finishes without returning False, it's a perfect match
        return True
```
