# Hash Map - aka dict

Hash Map (or Hash Table) is the general Computer Science concept.

dict (Dictionary) is Python's specific name for its built-in version of that concept.

for an unsorted array, a Hash Map is asymptotically the fastest possible way to solve the Two Sum problem

enumerate() = built-in function that allows you to loop through a list (or any other collection) and automatically keep track of both the index (the position) and the value (the item itself) at the same time.

enumerate = range(len())

```python
from typing import List

nums = [3,4,5,6]

target = 7

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i, num in enumerate(nums):
            n = target - num
            if n in seen:
                return [seen[n],i]
            seen[num] = i
        
        return []

s = Solution()

print(s.twoSum(nums, target))
```
